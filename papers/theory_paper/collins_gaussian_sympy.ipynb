```python
def complete_the_square(a, b, c, x):
    """
    Given the coefficients a, b, and c of a polynomial of the form:
        a*x^2 + b*x + c
    this function returns an equivalent expression of the form:
        a*(x + b/(2a))^2 + (c - b^2/(4a))
    i.e. the completed square plus leftover constants/terms.
    """
    square_part = a*(x + b/(2*a))**2
    leftover    = c - b**2/(4*a)
    
    return square_part + leftover
```

```python
import sympy as sp
from sympy import symbols, exp, sqrt, pi, I, integrate, oo, simplify


"""
Symbolically evaluates (using sp) the integral

    ∫ exp( -j π x1^2 / (q1  )
            -j π/(B λ)*(A x1^2 - 2 x1 x2 + D x2^2) )  dx1

and multiplies by the front factor sqrt(j/(B λ)).

Returns a simplified symbolic expression in terms of A,B,D,x2,q1,λ, , etc.
"""
x1, x0 = sp.symbols('x1 x0', real=True)
θ0 = sp.Symbol('θ_0', real=True)
x2 = sp.Symbol('x2', real=True)
A, B, D = sp.symbols('A B D', complex=True)
q1, q2, λ = sp.symbols('q_1 q_2, λ', complex=True)
j = I
# Define the exponent for plane wave solution:
exponent = (
         ((-j * 2 * sp.pi) / λ) * (x1 - x0) * θ0 - j * sp.pi * ((x1 - x0) ** 2)/(q1 * λ) +
         (-j * sp.pi) / (B * λ) * ( A * x1 ** 2  - 2 * x1 * x2)
)

prefactor = sqrt(j / (B * λ)) * sp.exp((-j * sp.pi * D * x2 ** 2) / (B * λ)) 

poly = sp.Poly(exponent, x1)

a = sp.simplify(poly.coeffs()[0])
b = sp.simplify(poly.coeffs()[1]/2)
c = sp.simplify(poly.coeffs()[2])

# a = (A / B + 1 / q1)
# b = -θ0 + x0 / q1 + x2 / B
display(poly)
display(a, b, c)
```
